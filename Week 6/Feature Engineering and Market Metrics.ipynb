# Feature Engineering and Market Metrics

```python
import pandas as pd
```

```python
date_cols = ["CloseDate", "PurchaseContractDate", "ListingContractDate"]
```

```python
listings = pd.read_csv("cleaned_listings.csv", parse_dates=date_cols)
sold = pd.read_csv("cleaned_sold.csv", parse_dates=date_cols)
```

### Calculating Key Metrics for Each Listing in the Datasets

##### Price Ratio for Negotiation Strength

```python
listings["PriceRatio"] = listings["ClosePrice"]/listings["OriginalListPrice"]
sold["PriceRatio"] = sold["ClosePrice"]/sold["OriginalListPrice"]
```

##### Price/SQR ft to Normalize Prices Across Sizes

```python
listings["PricePerSQRft"] = listings["ClosePrice"]/listings["LivingArea"]
sold["PricePerSQRft"] = sold["ClosePrice"]/sold["LivingArea"]
```

##### Close to Original Price to Capture Full Price Reduction History

```python
listings["ClosetoOriginalRatio"] = listings["ClosePrice"]/listings["OriginalListPrice"]
sold["ClosetoOriginalRatio"] = sold["ClosePrice"]/sold["OriginalListPrice"]
```

##### Listing to Contract Days to Measure Time from Listing to Accepted Offer

```python
listings["ListingtoContractDays"] = listings["PurchaseContractDate"] - listings["ListingContractDate"]
sold["ListingtoContractDays"] = sold["PurchaseContractDate"] - sold["ListingContractDate"]
```

##### Contract to Close Date to Measure Escrow and Closing Period Duration

```python
listings["ContractToCloseDate"] = listings["CloseDate"] - listings["PurchaseContractDate"]
sold["ContractToCloseDate"] = sold["CloseDate"] - sold["PurchaseContractDate"]
```

### Segment Analysis

Now, I want to understand how the metrics I just calculated compare among different groups of sold listings. This will help me see what trend analysis I want to perform in the future and capture in my Tableau visualizations.

##### Segmenting by Area

```python
sold.groupby("MLSAreaMajor")[["PriceRatio", "PricePerSQRft", "ClosetoOriginalRatio", "ListingtoContractDays", "ContractToCloseDate"]].median().reset_index()
```

##### Segmenting by Property Sub Type

```python
sold.groupby("PropertySubType")[["PriceRatio", "PricePerSQRft", "ClosetoOriginalRatio", "ListingtoContractDays", "ContractToCloseDate"]].median().reset_index()
```

##### Segmenting by County

```python
sold.groupby("CountyOrParish")[["PriceRatio", "PricePerSQRft", "ClosetoOriginalRatio", "ListingtoContractDays", "ContractToCloseDate"]].median().reset_index()
```

##### Segmenting by Listing Office

```python
sold.groupby("ListOfficeName")[["PriceRatio", "PricePerSQRft", "ClosetoOriginalRatio", "ListingtoContractDays", "ContractToCloseDate"]].median().reset_index()
```

##### Segmenting by Buyer Office

```python
sold.groupby("BuyerOfficeName")[["PriceRatio", "PricePerSQRft", "ClosetoOriginalRatio", "ListingtoContractDays", "ContractToCloseDate"]].median().reset_index()
```
