## Null Value Analysis

```python
sold.isnull().sum()
```

```python
perc_null = sold.isnull().sum()/len(sold)
perc_null
```

perc_null calculates the percentage of values that are null for each column in the sold dataset. 

```python
perc_null[perc_null > 0.9]
```

The above columns represent columns where over 90% of the values are null.
