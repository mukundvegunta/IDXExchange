# Data Cleaning and Preparation

### Loading in Finalized Datasets

```python
import pandas as pd
```

```python
listings = pd.read_csv("listings_with_rates.csv")
sold = pd.read_csv("sold_with_rates.csv")
```

### Converting Relevant Fields to Date-Time Format

```python
listings.columns
```

```python
listings["CloseDate"] = pd.to_datetime(listings["CloseDate"])
listings["PurchaseContractDate"] = pd.to_datetime(listings["PurchaseContractDate"])
listings["ListingContractDate"] = pd.to_datetime(listings["ListingContractDate"])
```

```python
sold.columns
```

```python
sold["CloseDate"] = pd.to_datetime(sold["CloseDate"])
sold["PurchaseContractDate"] = pd.to_datetime(sold["PurchaseContractDate"])
sold["ListingContractDate"] = pd.to_datetime(sold["ListingContractDate"])
```

### Validating Logical Order of Date Fields

#### Flagging in Listing Dataset

```python
listings["listings_after_close_flag"] = listings["ListingContractDate"] > listings["PurchaseContractDate"]
```

```python
listings[listings["listings_after_close_flag"] == True].head()
```

#### Making Sure that Flag Works as Intended

```python
listings[listings["ListingKey"] == 1154142014]["ListingContractDate"]
```

```python
listings[listings["ListingKey"] == 1154142014]["PurchaseContractDate"]
```

#### Continuing Check

```python
listings["purchase_after_close_flag"] = listings["CloseDate"] < listings["PurchaseContractDate"]
```

```python
listings["negative_timeline_flag"] = (listings["listings_after_close_flag"] == True) | (listings["purchase_after_close_flag"] == True)
```

#### Flagging in Sold Dataset

```python
sold["listings_after_close_flag"] = sold["ListingContractDate"] > sold["PurchaseContractDate"]
```

```python
sold["purchase_after_close_flag"] = sold["CloseDate"] < sold["PurchaseContractDate"]
```

```python
sold["negative_timeline_flag"] = (sold["listings_after_close_flag"] == True) | (sold["purchase_after_close_flag"] == True)
```

### Geographic Data Checks

```python
listings["no_coordinates"] = (listings["Longitude"].isna()) | (listings["Latitude"].isna())
sold["no_coordinates"] = (sold["Longitude"].isna()) | (sold["Latitude"].isna())
```

```python
listings[listings["no_coordinates"] == True]
```

```python
listings["sentinel_null_coordinates"] = (listings["Longitude"] == 0) | (listings["Latitude"] == 0)
sold["sentinel_null_coordinates"] = (sold["Longitude"] == 0) | (sold["Latitude"] == 0)
```

```python
listings["invalid_longitude"] = listings["Longitude"] > 0
sold["invalid_longitude"] = sold["Longitude"] > 0
```

California longitudes should always be negative.

### Dataset Adjustments

```python
len(listings)
```

#### Removing flagged entries

```python
listings_condition = (
    (listings["listings_after_close_flag"] == True) |
    (listings["purchase_after_close_flag"] == True) |
    (listings["negative_timeline_flag"] == True) |
    (listings["no_coordinates"] == True) |
    (listings["sentinel_null_coordinates"] == True) |
    (listings["invalid_longitude"] == True)
)
```

```python
listings = listings[~listings_condition]
```

#### Updated Row Count

```python
len(listings)
```

```python
len(sold)
```

#### Removing Flagged Entries

```python
sold_condition = (
    (sold["listings_after_close_flag"] == True) |
    (sold["purchase_after_close_flag"] == True) |
    (sold["negative_timeline_flag"] == True) |
    (sold["no_coordinates"] == True) |
    (sold["sentinel_null_coordinates"] == True) |
    (sold["invalid_longitude"] == True)
)
```

```python
sold = sold[~sold_condition]
```

#### Updated Row Count

```python
len(sold)
```

### Dropping Repetitive Columns

```python
listings = listings.loc[:, ~listings.columns.str.endswith(".1")]
```

```python
listings.columns
```

### Flagging Invalid Numeric Columns

```python
listings["invalid_numeric"] = listings[
    (listings["DaysOnMarket"] < 0) |
    (listings["LivingArea"] < 0) |
    (listings["ClosePrice"] < 0) |
    (listings["BathroomsTotalInteger"] < 0) |
    (listings["BedroomsTotal"] < 0) |
    (listings["LivingArea"] < 0)
].any(axis=1)
```

```python
sold["invalid_numeric"] = sold[
    (sold["DaysOnMarket"] < 0) |
    (sold["LivingArea"] < 0) |
    (sold["ClosePrice"] < 0) |
    (sold["BathroomsTotalInteger"] < 0) |
    (sold["BedroomsTotal"] < 0) |
    (sold["LivingArea"] < 0)
].any(axis=1)
```

```python
len(listings)
```

```python
listings_numeric_condition = listings["invalid_numeric"] == True
```

```python
listings = listings[~listings_numeric_condition]
```

```python
len(listings)
```

```python
len(sold)
```

```python
sold_numeric_condition = sold["invalid_numeric"] == True
```

```python
sold = sold[~sold_numeric_condition]
```

```python
len(sold)
```

Now that I made sure that entries with invalid entries or null values in important columns are removed, I am ready to export the datasets as csvs for further analysis.
