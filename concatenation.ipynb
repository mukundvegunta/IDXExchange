```python
import pandas as pd
```

# Listings Datasets

Loading listing datasets

```python
list1 = pd.read_csv("CRMLSListing202602.csv")
list2 = pd.read_csv("CRMLSListing202603.csv")
list3 = pd.read_csv("CRMLSListing202604.csv")
```

Checking size of each listing dataset and combined length before concatenation

```python
len(list1)
```

```python
len(list2)
```

```python
len(list3)
```

```python
len(list1) + len(list2) + len(list3)
```

Concatenating the listing datasets

```python
listings = pd.concat([list1, list2, list3])
```

```python
listings.head()
```

Ensuring that the length of the concatentated dataset matches the length of the indivudal datasets combined

```python
len(listings)
```

```python
listings.columns
```

Filtering for residential listings

```python
listings_residential = listings[listings["PropertyType"] == "Residential"]
```

```python
len(listings) - len(listings_residential)
```

There were 38,130 non-residential properties in the sold dataset that were filtered out.

```python
len(listings_residential)
```

The combined length for listed residential properties is 72,927.

# Sold Datasets

Repeating same process for sold datasets

```python
sold1 = pd.read_csv("CRMLSSold202602.csv")
sold2 = pd.read_csv("CRMLSSold202603.csv")
sold3 = pd.read_csv("CRMLSSold202604.csv")
```

```python
len(sold1) + len(sold2) + len(sold3)
```

```python
sold = pd.concat([sold1, sold2, sold3])
```

```python
len(sold)
```

```python
sold.columns
```

Ensuring that the date ranges of the combined dataset matches expectations.

```python
sold["CloseDate"].min()
```

```python
sold["CloseDate"].max()
```

Filtering for only residential properties.

```python
sold_residential = sold[sold["PropertyType"] == "Residential"]
```

```python
len(sold) - len(sold_residential)
```

There were 22,134 non-residential properties in the sold dataset that were filtered out.

```python
len(sold_residential)
```

The combined length for sold residential properties is 44,509.

## Saving Combined and Filtered Datasets as CSVs

```python
listings_residential.to_csv("listings_residential.csv", index=False)
```

```python
sold_residential.to_csv("sold_residential", index=False)
```
